### Problema abogado

```python
# ==============================================
# IA DE ABOGADO VIRTUAL
# Nivel: Principiante
# Uso de listas, if, elif, else y while True
# ==============================================

# Lista vacía al inicio
casos = []

print("===================================")
print("     IA DE ABOGADO VIRTUAL")
print("===================================")

# Ciclo principal
while True:

    print("\n------ MENÚ PRINCIPAL ------")
    print("1. Registrar caso")
    print("2. Ver casos")
    print("3. Buscar consejo legal")
    print("4. Eliminar último caso")
    print("5. Salir")

    opcion = input("Seleccione una opción: ")

    # ==========================================
    # REGISTRAR CASO
    # ==========================================
    if opcion == "1":

        nombre = input("Nombre del cliente: ")
        problema = input("Tipo de problema legal: ")

        casos.append([nombre, problema])

        print("\nCaso registrado correctamente.")
        print("La IA del abogado analizará el caso...")

        # Respuestas inteligentes simples
        if problema == "robo":
            print("Consejo IA: Reúna pruebas y denuncie.")

        elif problema == "divorcio":
            print("Consejo IA: Busque acuerdos pacíficos.")

        elif problema == "fraude":
            print("Consejo IA: Guarde mensajes y recibos.")

        elif problema == "despido":
            print("Consejo IA: Revise su contrato laboral.")

        else:
            print("Consejo IA: Consulte documentos legales.")

    # ==========================================
    # VER CASOS
    # ==========================================
    elif opcion == "2":

        if casos == []:
            print("\nNo hay casos registrados.")

        else:
            print("\nLISTA DE CASOS REGISTRADOS")
            print(casos)

    # ==========================================
    # BUSCAR CONSEJO LEGAL
    # ==========================================
    elif opcion == "3":

        consulta = input("\nEscriba el problema legal: ")

        if consulta == "robo":
            print("IA: Debe acudir al ministerio público.")

        elif consulta == "divorcio":
            print("IA: Puede iniciar un proceso familiar.")

        elif consulta == "fraude":
            print("IA: Reúna evidencias del engaño.")

        elif consulta == "despido":
            print("IA: Tiene derecho a liquidación.")

        elif consulta == "violencia":
            print("IA: Busque apoyo legal inmediato.")

        else:
            print("IA: No tengo suficiente información.")

    # ==========================================
    # ELIMINAR ÚLTIMO CASO
    # ==========================================
    elif opcion == "4":

        if casos == []:
            print("\nNo hay casos para eliminar.")

        else:
            eliminado = casos.pop()
            print("\nCaso eliminado correctamente.")
            print("Caso eliminado:", eliminado)

    # ==========================================
    # SALIR
    # ==========================================
    elif opcion == "5":

        print("\nGracias por usar la IA de abogado.")
        print("Sistema finalizado.")
        break

    # ==========================================
    # OPCIÓN INVÁLIDA
    # ==========================================
    else:
        print("\nOpción inválida. Intente otra vez.")
```
